```python
import yaml
from langgraph.graph import END, StateGraph
from langchain_core.runnables import RunnableLambda
from langchain_community.llms import HuggingFaceHub
from langchain.vectorstores import FAISS
from langchain.embeddings import HuggingFaceEmbeddings

# Load config
with open('graph_config.yaml', 'r') as f:
    config = yaml.safe_load(f)

# Initialize vector retriever
embedding_model = HuggingFaceEmbeddings(model_name='BAAI/bge-small-en')
vector_db = FAISS.load_local('leiden_guidelines_index', embedding_model)
retriever = vector_db.as_retriever()

# Node functions
def retrieve_context(state):
    docs = retriever.get_relevant_documents(state['question'])
    return {'context': '\n\n'.join([doc.page_content for doc in docs]), 'question': state['question'], 'audience': state.get('audience', 'default')}

def generate_answer_factory(model_id):
    llm = HuggingFaceHub(repo_id=model_id)
    def run(state):
        prompt = f'Document Excerpt:\n{state['context']}\n\nQuestion: {state['question']}'
        answer = llm.invoke(prompt)
        return {'answer': answer, **state}
    return run

def verify_answer_factory(model_id):
    llm = HuggingFaceHub(repo_id=model_id)
    def run(state):
        check_prompt = f'Is the following answer correctly formatted according to legal guidelines? Respond 'yes' or 'no':\n\n{state['answer']}'
        verdict = llm.invoke(check_prompt)
        if 'yes' in verdict.lower():
            return state
        else:
            return {'answer': 'Answer failed verification.', **state}
    return run

def adjust_answer_factory(model_id):
    llm = HuggingFaceHub(repo_id=model_id)
    def run(state):
        if state['audience'] == 'judge':
            adj_prompt = f'Rephrase this answer for a judge:\n\n{state['answer']}'
        elif state['audience'] == 'academic':
            adj_prompt = f'Rephrase this answer for an academic legal article:\n\n{state['answer']}'
        else:
            return state  # no adjustment
        adjusted = llm.invoke(adj_prompt)
        return {'answer': adjusted, **state}
    return run

# Graph construction from YAML
graph = StateGraph()

node_funcs = {}

for node in config['nodes']:
    if node['type'] == 'retriever':
        graph.add_node(node['name'], RunnableLambda(retrieve_context))
    elif node['type'] == 'llm':
        if 'verifier' in node['name']:
            graph.add_node(node['name'], RunnableLambda(verify_answer_factory(node['model'])))
        elif 'adjuster' in node['name']:
            graph.add_node(node['name'], RunnableLambda(adjust_answer_factory(node['model'])))
        else:
            graph.add_node(node['name'], RunnableLambda(generate_answer_factory(node['model'])))

# Add edges from config
for edge in config['edges']:
    graph.add_edge(edge['from'], edge['to'])

graph.set_entry_point('retriever')
graph_executor = graph.compile()

# Example invocation
result = graph_executor.invoke({
    'question': 'What language must videos be in to be admissible?',
    'audience': 'judge'
})

print(result['answer'])
```

```python
# from fuzzywuzzy import fuzz
import torch
from src.tokenizer import Tokenizer
from src.chat_message_history_builder import ChatMessageHistoryBuilder
import re
from transformers import AutoModelForCausalLM
#     'meta-llama/Llama-3.1-8B',


torch.cuda.empty_cache()
class LLM:

    def __init__(self, model_name: str, token: str):
        self.model_name = model_name

        device = 'cuda' if torch.cuda.is_available() else 'cpu'
        self.model = AutoModelForCausalLM.from_pretrained(
            model_name,
            token=HF_TOKEN,
            device_map='auto' if device == 'cuda' else None,
            torch_dtype=torch.float16,
            max_memory={0: '38GiB'}  # if 40GB available, leave headroom
        ).to(device)
        self.model.eval()
        self.tokenizer = Tokenizer(self.model_name, token=HF_TOKEN)

    def run_prompt(self, prompt):

        inputs = self.tokenizer.tokenize(prompt, max_length=1024)
        input_length = inputs['input_ids'].shape[1]

        if torch.cuda.is_available():
            logger.info('Using GPU.')
            self.model.to('cuda')
            inputs = {k: v.to('cuda') for k, v in inputs.items()}

        # TODO - increase the token limit to allow for more text
        eos_token_id = self.tokenizer.eos_token_id or self.tokenizer.convert_tokens_to_ids('<|end_of_text|>')

        generation_config = {
          'max_new_tokens': 500,
          'eos_token_id': eos_token_id,
          'no_repeat_ngram_size': 3,
          'repetition_penalty': 1.2,
          'pad_token_id': eos_token_id,
          'do_sample': False, # ensures the model stays strictly factual and consistent with the source material plus it always returns the consistency of the reponse.
          # 'num_beams': 3,  # Small beam width for better answers
          # temperature=0.3,
          # do_sample=True,

    #       'temperature': 0.0,
    # 'top_p': 1.0,
    # 'top_k': 50,


        }

        with torch.no_grad():
            outputs = self.model.generate(
              **inputs,
              **generation_config
            )
            new_tokens = outputs[0, input_length:]
            response = self.tokenizer.decode(new_tokens).strip()
            print(f'answer {response}')
            print(f'------')
        return response



llm = LLM('meta-llama/Llama-3.1-8B', token=HF_TOKEN)
```
